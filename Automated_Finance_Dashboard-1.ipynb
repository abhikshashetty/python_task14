```python
text_columns = [
    "Customer",
    "Category",
    "Product",
    "Payment_Method",
    "Region"
]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

print("Text columns cleaned successfully!")
```

```python
print("Negative Sales values:", (df["Sales"] < 0).sum())
print("Negative Profit values:", (df["Profit"] < 0).sum())
print("Negative Quantity values:", (df["Quantity"] < 0).sum())
```

```python
total_sales = df["Sales"].sum()
print("Total Sales Value:", total_sales)
```

```python
average_sales = df["Sales"].mean()
print("Average Sales:", average_sales)
```

```python
highest_sale = df["Sales"].max()
print("Highest Transaction Sales:", highest_sale)
```

```python
lowest_sale = df["Sales"].min()
print("Lowest Transaction Sales:", lowest_sale)
```

```python
total_transactions = len(df)
print("Total Number of Transactions:", total_transactions)
```

```python
total_profit = df["Profit"].sum()
print("Total Profit:", total_profit)
```

```python
category_summary = df.groupby("Category")["Sales"].agg(
    Number_of_Transactions="count",
    Total_Sales="sum",
    Average_Sales="mean"
).sort_values(
    "Total_Sales",
    ascending=False
)

category_summary
```

```python
region_summary = df.groupby("Region")["Sales"].agg(
    Number_of_Transactions="count",
    Total_Sales="sum",
    Average_Sales="mean"
).sort_values(
    "Total_Sales",
    ascending=False
)

region_summary
```

```python
payment_summary = df.groupby("Payment_Method")["Sales"].agg(
    Number_of_Transactions="count",
    Total_Sales="sum",
    Average_Sales="mean"
).sort_values(
    "Total_Sales",
    ascending=False
)

payment_summary
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(12, 6))
category_summary["Total_Sales"].plot(kind="bar")
plt.title("Total Sales Value by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
region_summary["Total_Sales"].plot(kind="bar")
plt.title("Total Sales Value by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
payment_summary["Average_Sales"].plot(kind="bar")
plt.title("Average Sales by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Average Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
daily_sales = df.groupby("Date")["Sales"].sum()

plt.figure(figsize=(12, 6))
daily_sales.plot(kind="line", marker="o")
plt.title("Sales Value by Date")
plt.xlabel("Date")
plt.ylabel("Total Sales Value")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df["Sales"], bins=30, edgecolor="black")
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Number of Transactions")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
top_products = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)
top_10_products = top_products.head(10)
top_10_products
```

```python
plt.figure(figsize=(12, 6))
top_10_products.plot(kind="bar")
plt.title("Top 10 Products by Sales Value")
plt.xlabel("Product")
plt.ylabel("Total Sales Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_profit = df.groupby("Category")["Profit"].sum()

plt.figure(figsize=(10, 5))
category_profit.sort_values().plot(kind="barh")
plt.title("Profit by Category")
plt.xlabel("Total Profit")
plt.ylabel("Category")
plt.tight_layout()
plt.show()
```

```python
payment_sales = df.groupby("Payment_Method")["Sales"].sum()

plt.figure(figsize=(8, 8))
plt.pie(
    payment_sales,
    labels=payment_sales.index,
    autopct="%1.1f%%"
)
plt.title("Sales Distribution by Payment Method")
plt.tight_layout()
plt.show()
```

```python
print("========== FINANCIAL DASHBOARD KPIs ==========")
print("Total Sales Value :", total_sales)
print("Total Profit :", total_profit)
print("Average Sales :", average_sales)
print("Highest Transaction Sales :", highest_sale)
print("Lowest Transaction Sales :", lowest_sale)
```

```python
top_category = category_summary["Total_Sales"].idxmax()
top_category_sales = category_summary["Total_Sales"].max()
print("Top Category:", top_category)
print("Sales Value:", top_category_sales)
```

```python
top_region = region_summary["Total_Sales"].idxmax()
top_region_sales = region_summary["Total_Sales"].max()
print("Top Region:", top_region)
print("Sales Value:", top_region_sales)
```

```python
top_profit_category = category_profit.idxmax()
top_profit_value = category_profit.max()
print("Most Profitable Category:", top_profit_category)
print("Profit:", top_profit_value)
```
